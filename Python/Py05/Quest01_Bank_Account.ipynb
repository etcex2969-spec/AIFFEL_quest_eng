# 🏦 파이썬 퀘스트 01 — 은행계좌 만들기

## 요구사항
1. `Account` 클래스: 은행 이름(SC은행), 예금주, 계좌번호(**3자리-2자리-6자리**), 잔고를 속성으로 가진다.
2. **클래스 변수** `account_count`로 지금까지 생성된 계좌 개수를 저장한다.
3. `deposit()`: 1원 이상만 입금 가능, **입금 5회가 될 때마다 잔고의 1%를 이자로 지급**한다.
4. `withdraw()`: 잔액 이하의 금액만 출금 가능하다.
5. `display_info()`: 은행, 예금주, 계좌번호, 잔고를 출력한다. (잔고는 천 단위 콤마)
6. `deposit_history()` / `withdraw_history()`: 입금·출금 내역을 출력한다.
7. 계좌 3개 이상을 만들고, 5회 이상 입금해 이자 지급을 확인한다.

```python
import random


class Account:
    """SC은행 계좌를 표현하는 클래스.

    Attributes:
        owner (str): 예금주 이름
        balance (int): 잔고
        account_number (str): 3자리-2자리-6자리 형식의 계좌번호
    """

    account_count = 0  # 클래스 변수: 생성된 계좌의 총 개수 (모든 인스턴스가 공유)

    def __init__(self, owner, balance):
        self.bank = "SC은행"
        self.owner = owner
        self.balance = balance
        self.deposit_count = 0   # 입금 횟수 (5회마다 이자 지급 확인용)
        self.deposit_log = []    # 입금 내역
        self.withdraw_log = []   # 출금 내역

        # 계좌번호 3자리-2자리-6자리를 랜덤으로 생성
        p1 = random.randint(100, 999)
        p2 = random.randint(10, 99)
        p3 = random.randint(100000, 999999)
        self.account_number = f"{p1}-{p2}-{p3}"

        Account.account_count += 1  # 계좌가 만들어질 때마다 클래스 변수 1 증가

    def deposit(self, amount):
        """입금. 1원 이상만 가능하며, 입금 5회마다 잔고의 1%를 이자로 지급한다."""
        if amount >= 1:
            self.balance += amount
            self.deposit_log.append(amount)
            self.deposit_count += 1
            if self.deposit_count % 5 == 0:  # 입금 5회마다 이자 지급
                interest = int(self.balance * 0.01)
                self.balance += interest
                print(f"{self.deposit_count}회 입금 완료! 이자 {interest:,}원이 지급되었습니다.")
            else:
                print(f"{amount:,}원 입금 완료")
        else:
            print("입금은 1원 이상만 가능합니다.")

    def withdraw(self, amount):
        """출금. 잔액 이하의 금액만 출금할 수 있다."""
        if amount <= self.balance:
            self.balance -= amount
            self.withdraw_log.append(amount)
            print(f"{amount:,}원 출금 완료")
        else:
            print("잔액이 부족합니다.")

    def display_info(self):
        """계좌 정보를 출력한다. 잔고는 천 단위마다 콤마로 표시한다."""
        print(f"은행: {self.bank} | 예금주: {self.owner} | "
              f"계좌번호: {self.account_number} | 잔고: {self.balance:,}원")

    def deposit_history(self):
        """입금 내역을 출력한다."""
        print(f"[{self.owner}] 입금 내역 ({len(self.deposit_log)}건)")
        if not self.deposit_log:
            print("  (기록 없음)")
        for i, amount in enumerate(self.deposit_log, 1):
            print(f"  {i}회차: {amount:,}원")

    def withdraw_history(self):
        """출금 내역을 출력한다."""
        print(f"[{self.owner}] 출금 내역 ({len(self.withdraw_log)}건)")
        if not self.withdraw_log:
            print("  (기록 없음)")
        for i, amount in enumerate(self.withdraw_log, 1):
            print(f"  {i}회차: {amount:,}원")
```

## 핵심 정리
- `account_count`는 **클래스 변수**라서 인스턴스가 아니라 `Account` 클래스 자체에 저장되고, 모든 계좌가 함께 공유합니다.
- 계좌번호는 `random.randint()` 세 번으로 3-2-6 자리를 만들고 f-string으로 연결했습니다.
- 이자 지급은 `deposit_count % 5 == 0` 조건 하나로 처리됩니다.
- `{value:,}` 포맷 지정자로 천 단위 콤마를 간단히 표현할 수 있습니다.

```python
# 출력 재현을 위해 랜덤 시드를 고정 (실제 사용 시에는 생략해도 됩니다)
random.seed(42)

# 1) 계좌 3개 생성
acc1 = Account("파이썬", 10000)
acc2 = Account("AIFFEL", 50000)
acc3 = Account("도담도담", 100000)
print(f"생성된 계좌 개수: {Account.account_count}개")
print()

# 2) 10회 입금 → 5회째, 10회째에 이자 지급 확인
for amount in [100, 200, 300, 400, 500, 600, 700, 800, 900, 1000]:
    acc1.deposit(amount)
print()

# 3) 예외 상황 확인
acc1.deposit(0)          # 1원 미만 → 입금 불가
acc1.withdraw(9999999)   # 잔액 초과 → 출금 불가
print()

# 4) 입금·출금 내역 출력
acc1.deposit_history()
acc1.withdraw_history()
print()

# 5) 계좌 정보 출력
acc1.display_info()
acc2.display_info()
acc3.display_info()
```

## 회고

### 배운 점
- 클래스 변수(`account_count`)는 객체마다 복사되는 것이 아니라 **클래스 하나에만 존재**하므로, 생성된 계좌 수를 세는 용도로 딱 맞았다.
- 5회마다 이자를 주는 규칙은 입금 횟수를 속성으로 기록해 두면 `% 5` 조건만으로 쉽게 구현된다.
- 입금·출금 내역을 리스트에 기록해 두면 `deposit_history()` / `withdraw_history()`에서 그대로 꺼내 보여주면 된다.

### 아쉬운 점 / 개선 아이디어
- 계좌번호를 랜덤으로 만들면 이론상 중복이 가능하다 → 실제 서비스라면 중복 검사가 필요하다.
- 내역을 리스트 대신 파일이나 데이터베이스에 저장하면 프로그램을 종료했다 켜도 남아 있을 것이다.
