# 📐 파이썬 퀘스트 02 — 다각형(사각형) 넓이 구하기

## 요구사항
1. `Square` 클래스 안에 세 도형의 넓이를 구하는 메서드를 만든다.
2. **직사각형**: 가로 × 세로
3. **평행사변형**: 밑변 × 높이
4. **사다리꼴**: (윗변 + 아랫변) × 높이 ÷ 2

```python
class Square:
    """사각형 3종(직사각형·평행사변형·사다리꼴)의 넓이를 계산하는 클래스."""

    def rec(self, width, height):
        """직사각형(rec-tangle)의 넓이: 가로 × 세로"""
        return width * height

    def par(self, base, height):
        """평행사변형(par-allelogram)의 넓이: 밑변 × 높이"""
        return base * height

    def trap(self, top, bottom, height):
        """사다리꼴(trap-ezoid)의 넓이: (윗변 + 아랫변) × 높이 ÷ 2"""
        return (top + bottom) * height / 2
```

## 핵심 정리
- 도형마다 클래스를 따로 만드는 대신, `Square` 하나에 `rec` · `par` · `trap` 세 메서드를 두면 넓이 계산 로직이 한 곳에 모입니다.
- 메서드 이름은 도형 영어 이름의 첫 음절(`rec-tangle`, `par-allelogram`, `trap-ezoid`)을 따왔고, 정확한 의미는 docstring으로 보강했습니다.
- `/` 연산은 **실수(float)**를 반환하기 때문에 사다리꼴 결과가 `75.0`처럼 표시됩니다. 정수가 필요하면 `int()`로 감싸면 됩니다.

```python
square = Square()  # 객체 생성

# 1) 직사각형: 가로 5, 세로 10
print(f"가로 5, 세로 10 직사각형의 넓이는 {square.rec(5, 10)}입니다!")

# 2) 평행사변형: 밑변 50, 높이 5
print(f"밑변 50, 높이 5인 평행사변형의 넓이는 {square.par(50, 5)}입니다!")

# 3) 사다리꼴: 윗변 10, 아랫변 20, 높이 5
print(f"윗변 10, 아랫변 20, 높이 5인 사다리꼴의 넓이는 {square.trap(10, 20, 5)}입니다!")
```

## 회고

### 배운 점
- 하나의 클래스에 여러 메서드를 두니 도형별 계산식이 한눈에 들어왔다.
- 사다리꼴 공식의 `÷ 2` 때문에 결과가 실수로 나온다. 출력 형식이 중요하면 f-string 포맷(`:.0f` 등)으로 다듬을 수 있다.
- docstring에 공식을 적어 두면 코드만 봐도 계산 원리를 알 수 있다.

### 아쉬운 점 / 개선 아이디어
- 가로·세로에 0이나 음수가 들어오면 검증하는 코드를 추가하면 더 안전하다.
- 도형이 늘어나면(예: 삼각형, 원) 부모 클래스 `Shape`를 두고 상속으로 확장하는 구조가 좋을 것이다.
