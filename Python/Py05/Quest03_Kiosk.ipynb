# ☕ 파이썬 퀘스트 03 — 키오스크 만들기

## 요구사항
1. 메뉴와 가격을 리스트로 저장하고, 번호와 함께 메뉴를 출력한다.
2. 음료 번호를 입력받아 주문한다.
3. **HOT(1) / ICE(2)** 온도를 입력받아 선택한 음료에 붙인다.
4. 추가 주문 여부(y/n)를 확인해 반복 주문을 받는다.
5. 주문 내역과 **총 결제 금액**(천 단위 콤마)을 출력한다.

```python
class Kiosk:
    """음료를 주문하는 간단한 키오스크 클래스."""

    def __init__(self):
        self.menu = ["아메리카노", "라떼", "모카", "유자차", "꿀차", "녹차"]
        self.price = [2000, 3000, 3000, 2500, 2500, 3000]
        self.order_menu = []   # 주문한 음료 이름 (온도 포함)
        self.order_price = []  # 주문한 음료 가격
        self.order_sum = 0     # 총 결제 금액

    def menu_print(self):
        """1. 메뉴 출력: 번호, 이름, 가격을 한 줄씩 보여준다."""
        print("======== 음료 메뉴 ========")
        for i in range(len(self.menu)):
            print(f"{i + 1}. {self.menu[i]} : {self.price[i]:,}원")

    def menu_select(self):
        """2~3. 음료 1개 주문: 번호 선택 → 온도 선택 → 주문 목록에 추가."""
        n = 0
        while n < 1 or n > len(self.menu):  # 메뉴 번호를 벗어나면 다시 입력
            n = int(input("음료 번호를 입력하세요: "))
            if n < 1 or n > len(self.menu):
                print("메뉴에 있는 번호로 다시 입력하세요.")

        t = 0
        while t != 1 and t != 2:  # 1 또는 2를 입력할 때까지 반복
            t = int(input("HOT 음료는 1을, ICE 음료는 2를 입력하세요: "))
            if t == 1:
                self.temp = "HOT"
            elif t == 2:
                self.temp = "ICE"
            else:
                print("1 또는 2를 다시 입력하세요.")

        name = f"{self.temp} {self.menu[n - 1]}"  # 예: "HOT 아메리카노"
        self.order_menu.append(name)
        self.order_price.append(self.price[n - 1])
        self.order_sum += self.price[n - 1]
        print(f"{name} : {self.price[n - 1]:,}원 선택")

    def table(self):
        """5. 주문 내역과 총 결제 금액을 출력한다."""
        print()
        print("======== 주문 내역 ========")
        for name, p in zip(self.order_menu, self.order_price):
            print(f"  {name} : {p:,}원")
        print(f"총 결제 금액: {self.order_sum:,}원")

    def run(self):
        """전체 주문 흐름: 메뉴 출력 → (주문 → 추가 주문 y/n) 반복 → 내역 출력."""
        self.menu_print()
        while True:
            print()
            self.menu_select()
            answer = ""
            while answer not in ("y", "n"):  # y 또는 n을 입력할 때까지 반복
                answer = input("추가 주문을 하시겠습니까? (y/n): ")
            if answer == "n":
                break
        self.table()
```

## 핵심 정리
- `menu`(이름)와 `price`(가격)를 **같은 인덱스로 연결된 두 리스트**로 관리합니다.
- `menu_select()`는 음료 하나를 주문하는 단위 메서드, `run()`은 전체 흐름(반복 주문)을 담당합니다. 역할을 나누면 코드를 읽고 고치기 쉽습니다.
- 잘못된 입력(메뉴에 없는 번호, 1·2가 아닌 온도, y/n이 아닌 답변)은 `while` 반복문으로 다시 묻는 방식으로 처리합니다.
- 주문 이름은 f-string으로 온도와 음료를 붙여 만듭니다: `f"{self.temp} {self.menu[n - 1]}"`

```python
# input()을 미리 준비한 답변으로 대체해 자동 시연합니다.
# (실제 키오스크는 키보드 입력을 받습니다: 1번 아메리카노 HOT → 추가 주문 y →
#  4번 유자차 ICE → 추가 주문 n)
answers = iter(["1", "1", "y", "4", "2", "n"])
input = lambda prompt="": next(answers)

kiosk = Kiosk()
kiosk.run()

del input  # 시연이 끝나면 원래 input()으로 되돌림
```

## 회고

### 배운 점
- `input()`이 돌려주는 값은 **문자열**이므로 `int()`로 변환해야 숫자 비교가 가능하다.
- 주문을 리스트에 `append()`로 차곡차곡 쌓고 마지막에 한 번에 출력하는 구조가 깔끔했다.
- `zip()`으로 이름 리스트와 가격 리스트를 나란히 꺼내 출력할 수 있었다.

### 아쉬운 점 / 개선 아이디어
- 숫자가 아닌 값(예: `"a"`)을 입력하면 `ValueError`가 발생한다 → `try/except`로 처리하면 더 견고해진다.
- 메뉴와 가격을 두 리스트 대신 딕셔너리 `{"아메리카노": 2000, ...}` 하나로 관리하면 인덱스 실수 가능성이 줄어든다.
